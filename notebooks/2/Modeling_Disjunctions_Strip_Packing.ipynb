### Problem Statement

![Four pairwise no-overlap certificates using bottom-left coordinates.](https://raw.githubusercontent.com/ndcbe/optimization/bd9839ba4c3c0d23326aab67eada00c4196a0105/media/figures/lm-packing-separation.png)


Use the **bottom-left corner** $(x_i,y_i)$ of each rectangle, as in the code below. Length $L_i$ runs along $x$ and height $H_i$ runs along $y$. The fixed strip width is $W=10$, and rotation is not allowed.

| Rectangle $i$ | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| --- | --- | --- | --- | --- | --- | --- | --- | --- |
| Length $L_i$ | 4 | 3 | 2 | 2 | 3 | 3 | 4 | 4 |
| Height $H_i$ | 3 | 3 | 2 | 2 | 3 | 5 | 7 | 7 |

Let $N=\{0,\ldots,7\}$, $\mathcal{P}=\{(i,j):i,j\in N,\ i<j\}$, and $B=\sum_iL_i=25$. The actual implemented model is

$$\begin{aligned}
\min\quad &\ell\\
\text{subject to}\quad &\ell\ge x_i+L_i,\quad i\in N,\\
&0\le x_i\le B,\quad 0\le y_i\le W-H_i,\quad i\in N,\\
&\ell\ge0,\\
&\begin{bmatrix}Y^1_{ij}\\x_i+L_i\le x_j\end{bmatrix}\lor
\begin{bmatrix}Y^2_{ij}\\x_j+L_j\le x_i\end{bmatrix}\lor
\begin{bmatrix}Y^3_{ij}\\y_i+H_i\le y_j\end{bmatrix}\lor
\begin{bmatrix}Y^4_{ij}\\y_j+H_j\le y_i\end{bmatrix},\quad(i,j)\in\mathcal{P}.
\end{aligned}$$

Exactly one **certificate indicator** is selected for each pair. The first two alternatives put one rectangle to the left of the other; the last two put one below the other. Diagonally separated rectangles may satisfy both a horizontal and a vertical inequality. This remains feasible because an unselected indicator does not enforce the opposite of its inequality.

There are $\binom82=28$ disjunctions and $4\binom82=112$ binary indicators after reformulation. The declared $x_i\le25$ is the code's bound, not a tighter heuristic bound $x_i\le B-L_i$. The model does not declare an upper bound on $\ell$; that variable appears only in global rows and the objective, so its absence does not obstruct hull's bound requirement for variables **inside disjuncts**. A feasible length-25 placement exists by putting all rectangles side by side, so the finite coordinate bound preserves at least one optimum.

The original MINLP library formulation uses an upper-left convention in its exposition. Changing that convention requires changing the signs and bounds consistently; the equations above match this notebook's bottom-left implementation.

### Define model in Pyomo with GDP

First we will define the model for the Strip Packing Problem in Pyomo. 

`````python
"""
Instead of using
# import pyomo.environ as pyo
We can import specific functions/objects
"""

from pyomo.environ import (
    check_optimal_termination,
    ConcreteModel,
    NonNegativeReals,
    Objective,
    Param,
    Set,
    SolverFactory,
    TransformationFactory,
    Var,
    value,
)
`````

`````python
# Strip-packing example from http://minlp.org/library/lib.php?lib=GDP

# This model packs a set of rectangles without rotation or overlap within a
# strip of a given width, minimizing the length of the strip.


def create_model():
    """
    Build the strip packing problem model.

    Return:
    model: Pyomo model

    """

    ## Model
    model = ConcreteModel(name="Rectangles strip packing")

    ## Sets
    model.rectangles = Set(ordered=True, initialize=[0, 1, 2, 3, 4, 5, 6, 7])

    ## Parameters
    # Extent of each rectangle across the width of the strip (y direction)
    model.rect_width = Param(
        model.rectangles, initialize={0: 3, 1: 3, 2: 2, 3: 2, 4: 3, 5: 5, 6: 7, 7: 7}
    )

    # Parameter indexed by each rectangle.
    # rect_length is the extent along the strip (the x direction);
    # rect_width is the extent across the strip (the y direction).
    model.rect_length = Param(
        model.rectangles, initialize={0: 4, 1: 3, 2: 2, 3: 2, 4: 3, 5: 3, 6: 4, 7: 4}
    )

    model.strip_width = Param(initialize=10, doc="Available width of the strip")

    # Upperbound on length (default is sum of lengths of rectangles)
    model.max_length = Param(
        initialize=sum(model.rect_length[i] for i in model.rectangles),
        doc="Maximum length of the strip (if all rectangles were arranged "
        "lengthwise)",
    )

    ## Variables
    # x (length) and y (width) coordinates of each of the rectangles
    model.x = Var(
        model.rectangles,
        bounds=(0, model.max_length),
        doc="Rectangle corner x-position (position across length)",
    )

    # Width bounds
    def w_bounds(b, i):
        return (0, b.strip_width - b.rect_width[i])

    model.y = Var(
        model.rectangles,
        bounds=w_bounds,
        doc="Rectangle corner y-position (position down width)",
    )
    # Strip length
    model.strip_length = Var(domain=NonNegativeReals, doc="Length of strip required.")

    # Rectangle conflicts
    def rec_pairs_filter(b, i, j):
        return i < j

    model.overlap_pairs = Set(
        initialize=model.rectangles * model.rectangles,
        dimen=2,
        filter=rec_pairs_filter,
        doc="Set of possible rectangle conflicts",
    )

    ## Constraints
    @model.Constraint(model.rectangles)
    def strip_ends_after_last_rec(b, i):
        return b.strip_length >= b.x[i] + b.rect_length[i]

    ## Objective
    model.total_length = Objective(expr=model.strip_length, doc="Minimize length")

    ## Insert the no-overlap disjunctions here!

    # Add your solution here

    return model
`````

### Extract the Big-M placement

Store the solved coordinates and check containment, all pairwise separations, and the selected certificates before replacing the model for the hull solve. These checks use the actual solution, not coordinates from an earlier solver run.

`````python
import json
import math
import pyomo.environ as pyo
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pyomo

def extract_packing(m, method):
    records = [{"id": int(i), "length": pyo.value(m.rect_length[i]),
                "height": pyo.value(m.rect_width[i]),
                "x": pyo.value(m.x[i]), "y": pyo.value(m.y[i])}
               for i in m.rectangles]
    length, width = pyo.value(m.strip_length), pyo.value(m.strip_width)
    tol = 1e-6
    for r in records:
        assert r["x"] >= -tol and r["y"] >= -tol
        assert r["x"] + r["length"] <= length + tol
        assert r["y"] + r["height"] <= width + tol
    certificates = []
    for i, j in m.overlap_pairs:
        a, b = records[i], records[j]
        residuals = [a["x"]+a["length"]-b["x"], b["x"]+b["length"]-a["x"],
                     a["y"]+a["height"]-b["y"], b["y"]+b["height"]-a["y"]]
        assert min(residuals) <= tol
        selected = [k for k,d in enumerate(m.no_overlap[i,j].disjuncts)
                    if pyo.value(d.binary_indicator_var) > 0.5]
        assert len(selected) == 1 and residuals[selected[0]] <= tol
        certificates.append({"pair": [int(i),int(j)], "selected": selected[0],
                             "separation_residuals": residuals})
    variables = list(m.component_data_objects(pyo.Var, active=True))
    rows = list(m.component_data_objects(pyo.Constraint, active=True))
    return {"method": method, "length": length, "width": width, "rectangles": records,
            "certificates": certificates,
            "size": {"binary_variables": sum(v.is_binary() for v in variables),
                     "continuous_variables": sum(not v.is_binary() for v in variables),
                     "equalities": sum(c.equality for c in rows),
                     "inequalities": sum(not c.equality for c in rows)}}

packing_bigm = extract_packing(model, "gdp.bigm")
print(json.dumps(packing_bigm, indent=2))
`````

## Discussion Questions

1. What does the indexed disjunction decorator create for one rectangle pair?
2. How many binary indicators are introduced for eight rectangles? Is that number a count of continuous degrees of freedom?
3. Which coordinate and bound would change if rectangle rotation were permitted?
4. Do the two methods need to return identical optimal coordinates?
5. Why does exactly-one certificate selection allow diagonal separation?

```{dropdown} Answers
1. Four alternatives, each with one separating inequality. The decorator repeats that four-term structure for each pair.
2. There are 28 pairs and four indicators per pair, giving 112 binaries. This component count is not a Jacobian-rank or degrees-of-freedom calculation.
3. Rotation would interchange a rectangle's length and height; a new orientation decision would have to enter both the separating inequalities and the containment bounds. The present model fixes orientation.
4. No. Both methods return the optimal objective 11, but multiple placements can attain it. Their coordinates and solver work can differ. Neither relative iteration count nor speed is guaranteed by the formulation names.
5. Exactly one selected indicator requires its inequality. A false indicator leaves its inequality unrestricted; that inequality may still happen to be true.
```

## Certify the optimal length

The total rectangle area is $\sum_i L_iH_i=109$, so containment in a width-10 strip gives $\ell\ge109/10=10.9$. A stronger geometric bound is available here. Rectangles 5, 6, and 7 have heights 5, 7, and 7. Every pair has height sum greater than 10, so no pair can be separated vertically inside this strip. Their horizontal intervals must therefore be pairwise disjoint. Their lengths sum to $3+4+4=11$, implying $\ell\ge11$.

Each extracted solution is independently checked above for containment and all 28 no-overlap conditions. Each has length 11, attaining the bound, so both are globally optimal for the stated fixed-orientation data. Their one unit of unused area is $10(11)-109$. This argument certifies the objective without claiming unique coordinates or relying solely on a solver status.

`````python
area = sum(r["length"]*r["height"] for r in packing_bigm["rectangles"])
tall = [r for r in packing_bigm["rectangles"] if r["id"] in (5,6,7)]
assert all(a["height"]+b["height"] > packing_bigm["width"]
           for k,a in enumerate(tall) for b in tall[k+1:])
length_bound = sum(r["length"] for r in tall)
assert area == 109 and length_bound == 11
assert all(abs(p["length"]-length_bound) < 1e-6 for p in (packing_bigm,packing_hull))
results = {"packings": [packing_bigm,packing_hull], "area": area,
           "area_lower_bound": area/packing_bigm["width"],
           "height_conflict_ids": [5,6,7], "certified_lower_bound": length_bound}
nb_path = "notebooks/2-dev/Modeling_Disjunctions_Strip_Packing.ipynb"
helper.save_results("strip-packing-solution", results, notebook=nb_path,
    source_tag="handout:strip-packing-model",
    source_cells=[(nb_path,t) for t in ("l4:strip-bigm-transform", "l4:strip-bigm-solve",
                  "l4:strip-hull-solve", "l4:strip-extract-bigm", "l4:strip-extract-hull",
                  "l4:strip-archive")],
    solver=f"{milp_solver}; Pyomo {pyomo.__version__}",
    description="Actual Big-M and hull fixed-orientation placements; independent feasibility checks and exact length-11 certificate.")
print(f"Area bound = {area/10:g}; height-conflict bound = {length_bound:g}; both feasible lengths = 11.")
`````

`````python
def plot_packings(results):
    from matplotlib.patches import Rectangle
    # Hatch identities follow figures/plots/_house.py, with two extra patterns.
    hatches = ("///", "\\\\\\", "...", "xxx", "|||", "---", "ooo", "+++")
    colors = ("#0072B2", "#D55E00", "#009E73", "#CC79A7", "#56B4E9", "#E69F00", "#F0E442", "#0072B2")
    fig, axes = plt.subplots(1,2,figsize=(10.8,5.1),sharex=True,sharey=True)
    for ax, packing in zip(axes, results["packings"]):
        for r in packing["rectangles"]:
            i = r["id"]
            ax.add_patch(Rectangle((r["x"],r["y"]),r["length"],r["height"],
                         facecolor=colors[i],alpha=0.22,edgecolor="black",hatch=hatches[i],
                         linewidth=2.4 if i in results["height_conflict_ids"] else 1.2))
            # An opaque border and label keep every rectangle distinct in monochrome.
            ax.add_patch(Rectangle((r["x"],r["y"]),r["length"],r["height"],
                         fill=False,edgecolor="black",linewidth=2.4 if i in results["height_conflict_ids"] else 1.2))
            ax.text(r["x"]+r["length"]/2,r["y"]+r["height"]/2,str(i),
                    ha="center",va="center",fontsize=14,fontweight="bold",
                    bbox=dict(facecolor="white",edgecolor="none",pad=1.2))
        ax.set(xlim=(-0.25,11.25),ylim=(-0.25,10.25),xlabel="Along strip, $x$",
               title=f'{packing["method"]}: length {packing["length"]:.4g}')
        ax.set_xticks([0,3,6,9,11]);ax.set_yticks([0,2,4,6,8,10]);ax.set_aspect("equal")
        for spine in ax.spines.values(): spine.set_color("black")
        ax.tick_params(colors="black")
    axes[0].set_ylabel("Across strip, $y$")
    fig.tight_layout()
    return fig

fig = plot_packings(results)
helper.save_figure(fig, "strip-packing-solution")
`````
