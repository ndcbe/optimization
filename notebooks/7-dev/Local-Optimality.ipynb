# Local Optimality Conditions


**Reference** Section 4.2 in Biegler (2010)

## Unconstrained comparison

For twice continuously differentiable objectives, a local unconstrained minimum has zero gradient and positive-semidefinite Hessian. These are necessary conditions; they do not classify every stationary point. The definitions, derivative tests and proofs have their existing home in [Unconstrained Optimality Conditions](../6/Optimality.ipynb). Here the feasible neighborhood and constraint reactions change the test.


## Balance of Forces Interpretation

Here is a [helpful handout](../../handouts/OptimalityConditionsSummary.pdf).

(ci-ref-fence)=
Consider $\min f(\mathbf{x})$ subject to $g(\mathbf{x})\le0$. The objective is a landscape; its downhill force is $-\nabla f$. The fence force is $-u\nabla g$, with $u\ge0$, pointing toward the feasible side.

![The source-faithful fence picture with corrected gradient normals.](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/ci-fence-gradients.png)

The drawn arrows are gradients, pointing toward increasing values. At a regular constrained minimum with an active pushing fence,

$$\nabla f(\mathbf{x}^*)+u^*\nabla g(\mathbf{x}^*)=0.$$

The objective contour is tangent to the fence. Opposite gradient directions do not require equal lengths: the multiplier supplies the scale. Rescaling a constraint changes the numerical multiplier, so raw arrow lengths are not force magnitudes.

Source: Biegler (2010), Figures 4.5–4.6, pp.68–69. The handout and website use this same prepared drawing.


(ci-ref-rail)=
An equality $h(\mathbf{x})=0$ acts as a rail. Its force $-v\nabla h$ can point either way, because $v$ is unrestricted.

![The rail and fence with their corrected gradient normals.](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/ci-rail-gradients.png)

Balance is $\nabla f+u\nabla g+v\nabla h=0$. For this illustrative drawing the normalized gradients are approximately

$$\nabla f=(0.9825,0.1861)^T,\quad
\nabla g=(-0.8587,-0.5125)^T,\quad
\nabla h=(-0.8480,0.5300)^T.$$

They give $u\approx0.76265>0$, $v\approx0.38634$. The direction $\mathbf p=(0.5300,0.8480)^T$ is tangent to the rail, enters the feasible side of the fence, and raises the objective to first order. Its negative crosses the fence. Thus there is feasible motion, but no nonzero **critical** direction at this intersection. “No allowable directions” in a curvature discussion refers to the critical cone, not all feasible motion.

Removing the rail permits a downhill direction along the fence toward its tangency minimum. Keeping the rail removes that descent direction. Source: Biegler (2010), Figure 4.7 and Eq. (4.9), pp.70–71. The later numerical example uses its own objective and gradients; these normalized drawing vectors are not substituted into that model.


(ci-ref-kkt)=
## Karush-Kuhn-Tucker (KKT) necessary conditions

Use row-oriented Jacobians $\mathbf J_g\in\mathbb R^{r\times n}$ and $\mathbf J_h\in\mathbb R^{m\times n}$ and

$$\mathcal L(\mathbf x,\mathbf u,\mathbf v)=f(\mathbf x)+
\mathbf g(\mathbf x)^T\mathbf u+\mathbf h(\mathbf x)^T\mathbf v.$$

Assume $C^1$ data on an open neighborhood of a feasible local minimum. If LICQ holds there, the following necessary multiplier conditions apply:

$$\begin{aligned}
\nabla_{\mathbf x}\mathcal L&=\nabla f+\mathbf J_g^T\mathbf u+\mathbf J_h^T\mathbf v=0,\\
\mathbf g&\le0,\qquad\mathbf h=0,\qquad\mathbf u\ge0,\\
g_i u_i&=0\quad\text{for every }i.
\end{aligned}$$

Evaluate these at the candidate. Equality multipliers have no sign restriction. Dot-product complementarity $\mathbf g^T\mathbf u=0$ is equivalent to the component conditions because every summand is nonpositive. The prepared product-vector diagram compares nonpositive terms with a zero vector. If any term is strictly negative, no other nonpositive term can cancel it to make the sum zero.

![Nonpositive products and their zero-sum comparison.](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/ci-complementarity.png)

Inactive constraints have $g_i<0$, hence $u_i=0$. Active constraints have $g_i=0$ and may have a positive multiplier (strongly active) or zero multiplier (weakly active).

**Regularity is a hypothesis.** LICQ means the equality gradients and active inequality gradients are linearly independent. It is one sufficient qualification for the first-order necessary conditions. For $\min x$ subject to $x^2=0$, the only feasible point is a strict global minimum, but $1+v(2x)=0$ cannot hold at zero for any multiplier. The constraint gradient is zero. See [Constraint Qualifications](Constraint-Qualifications.ipynb) for the full theory.

(ci-ref-critical)=
**Feasible and critical directions differ.** The linearized feasible cone imposes $\mathbf J_h\mathbf p=0$ and $\nabla g_i^T\mathbf p\le0$ on active inequalities. At a KKT point,

$$\nabla f^T\mathbf p=-\sum_iu_i\nabla g_i^T\mathbf p\ge0.$$

Critical directions have zero first-order objective change. Since every term on the right is nonnegative, they are exactly the directions satisfying

$$\begin{aligned}
\mathbf J_h\mathbf p&=0,\\
\nabla g_i^T\mathbf p&=0&&g_i=0,\ u_i>0,\\
\nabla g_i^T\mathbf p&\le0&&g_i=0,\ u_i=0.
\end{aligned}$$

A positive multiplier does not prevent every feasible inward move; it makes that move raise the objective to first order. Such a direction is not critical. These are first-order tangent tests; nonlinear constraints need not remain feasible along an entire straight step.

![Feasible directions versus critical directions for a pushing or touching fence.](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/ci-critical-directions.png)

For $g=-x_2$ at the origin, $f=x_1^2+x_2$ has $u=1$ and critical directions $p_2=0$. Its feasible directions with $p_2>0$ already raise the objective. For $f=x_1^2+x_2^2$, $u=0$ and every linearized feasible direction $p_2\ge0$ is critical.

With $C^2$ data on an open neighborhood and LICQ at a local minimum, second-order necessity gives

$$\mathbf p^T\nabla_{\mathbf{xx}}^2\mathcal L\,\mathbf p\ge0
\quad\text{for every critical direction }\mathbf p.$$

Together with KKT, strict positivity on every nonzero critical direction is sufficient for a strict local minimum. The two examples satisfy it on their respective cones. If the critical cone is $\{0\}$, the strict condition is vacuous; feasible noncritical directions are controlled by their positive first-order change. Full theorem/proof material has its existing home in [Second-Order Conditions](Second-Order.ipynb).

Sources: Biegler (2010), Eqs. (4.9)–(4.12), pp.70–71; Theorems 4.14–4.15, pp.77–78; critical cone in §4.3.3, p.79; Theorems 4.17–4.18, pp.80–81. The constraint-qualification counterexample follows directly from its singleton feasible set and zero constraint derivative.


## Kinematic Interpretation via Example

Consider the following two dimensional optimization problem:

$$
\begin{align} \min_{x_1,x_2} \quad & f(\mathbf{x}) := x_1^2 - 4 x_1 + \frac{3}{2} x_2^2 -7x_2 + x_1 x_2 + 9 - \mathrm{ln}(x_1) - \mathrm{ln}(x_2) \\
\mathrm{s.t.} \quad & g(\mathbf{x}) := 4 - x_1 x_2 \leq 0 \\
 & h(\mathbf{x}) := 2 x_1 - x_2 = 0
\end{align}
$$

```python
import sys
if "google.colab" in sys.modules:
    import urllib.request
    urllib.request.urlretrieve(
        "https://raw.githubusercontent.com/ndcbe/optimization/main/notebooks/helper.py",
        "helper.py",
    )
    import helper
    helper.easy_install()
else:
    sys.path.insert(0, "..")
    import helper

helper.set_plotting_style()
import numpy as np
import matplotlib.pyplot as plt
import pyomo.environ as pyo
```

### Define Function for Visualization

```python
## Objective function
def f(x):
    """Evaluate the two-variable objective function."""
    return (
        x[0] ** 2
        - 4 * x[0]
        + 1.5 * x[1] ** 2
        - 7 * x[1]
        + x[0] * x[1]
        + 9
        - np.log(x[0])
        - np.log(x[1])
    )


## Gradient of objective f(x)
def df(x):
    """Evaluate the objective gradient."""
    return np.array((2 * x[0] - 4 + x[1] - 1 / x[0], 3 * x[1] - 7 + x[0] - 1 / x[1]))


## Gradient of inequality constraint g(x)
def dg(x):
    """Evaluate the inequality-constraint gradient."""
    return np.array((-x[1], -x[0]))


## Gradient of equality constraint h(x)
def dh(x):
    """Evaluate the equality-constraint gradient."""
    return np.array([2, -1])
```

### Define function to solve optimization problem with Pyomo

```python
def solve_opt(consider_g, consider_h):
    """Solve the example with the selected constraints enabled."""

    ## Create concrete Pyomo model
    m = pyo.ConcreteModel()

    ## Declare variables with initial values
    m.x1 = pyo.Var(bounds=(0, 100), initialize=10)
    m.x2 = pyo.Var(bounds=(0, 100), initialize=10)

    ## Declare objective
    m.OBJ = pyo.Objective(
        expr=m.x1**2
        - 4 * m.x1
        + 1.5 * m.x2**2
        - 7 * m.x2
        + m.x1 * m.x2
        + 9
        - pyo.log(m.x1)
        - pyo.log(m.x2),
        sense=pyo.minimize,
    )

    if consider_g:
        ## Add inequality constraint
        m.con1 = pyo.Constraint(expr=4 - m.x1 * m.x2 <= 0)

    if consider_h:
        ## Add equality constraint
        m.con2 = pyo.Constraint(expr=2 * m.x1 - m.x2 == 0)

    ## Specify IPOPT as solver
    solver = pyo.SolverFactory("ipopt")

    ## Solve the model
    results = solver.solve(m, tee=True)
    assert pyo.check_optimal_termination(results), (
        f"Solve failed: status={results.solver.status}, "
        f"termination={results.solver.termination_condition}"
    )

    ## Return the solution
    return [pyo.value(m.x1), pyo.value(m.x2)]
```

### Three constraint cases

The three solves below feed the shared prepared figure after Take 3. The complete visualization, plotting implementation and archived results have one source in this notebook.


### Take 1: Unconstrained

```python
kkt_solutions = {}
xsln = solve_opt(False, False)
kkt_solutions["unconstrained"] = list(xsln)
print("Primal point:", np.asarray(xsln))
print("Objective:", f(xsln))
```

### Take 2. With $g(\mathbf{x}) \leq 0$

```python
xsln = solve_opt(True, False)
kkt_solutions["inequality"] = list(xsln)
print("Primal point:", np.asarray(xsln))
print("Objective:", f(xsln))
```

### Take 3. With $g(\mathbf{x}) \leq 0$ and $h(\mathbf{x}) = 0$

```python
xsln = solve_opt(True, True)
kkt_solutions["both"] = list(xsln)
print("Primal point:", np.asarray(xsln))
print("Objective:", f(xsln))
```

### Discussion

Why are the gradient vectors not the same length? I thought the forces were balanced...

### Analysis without Constraints

For the unconstrained problem, stationarity requires

$$
\nabla f(\mathbf{x}^*) =
\begin{bmatrix}
2x_1-4+x_2-1/x_1 \\
3x_2-7+x_1-1/x_2
\end{bmatrix}_{\mathbf{x}=\mathbf{x}^*}=\mathbf{0}.
$$

Numerically, $\mathbf{x}^*=(1.3475,\,2.0470)$ and $f(\mathbf{x}^*)=-0.8742$. Moreover,

$$
\nabla^2 f(\mathbf{x}^*) =
\begin{bmatrix}
2+1/(x_1^*)^2 & 1 \\
1 & 3+1/(x_2^*)^2
\end{bmatrix}
=
\begin{bmatrix}
2.5507 & 1 \\
1 & 3.2387
\end{bmatrix}
\succ 0,
$$

so $\mathbf{x}^*$ is a strict local minimizer.

### Analysis with Constraints

For the constrained problem, define

$$\mathcal{L}(\mathbf{x},u,v)=f(\mathbf{x})+u\,g(\mathbf{x})+v\,h(\mathbf{x}).$$

The exact point is $(\sqrt2,2\sqrt2)^T$. Solving the KKT conditions gives the rounded values

$$
\mathbf{x}^*=(1.4142,\,2.8284),\qquad u^*=1.068,\qquad v^*=1.0355,
$$

with $g(\mathbf{x}^*)=0$, $h(\mathbf{x}^*)=0$, and $f(\mathbf{x}^*)=0.1579$. The Lagrangian Hessian is

$$
\nabla^2_{\mathbf{xx}}\mathcal{L}(\mathbf{x}^*,u^*,v^*)=
\begin{bmatrix}
2+1/(x_1^*)^2 & 1-u^* \\
1-u^* & 3+1/(x_2^*)^2
\end{bmatrix}
=
\begin{bmatrix}
2.5 & -0.068 \\
-0.068 & 3.125
\end{bmatrix}.
$$

The exact multipliers are $u^*=(17-9\sqrt2)/4>0$ and $v^*=(5/2)(\sqrt2-1)$. The active inequality and equality gradients are independent. Since the inequality multiplier is positive, a critical direction must be tangent to both constraints; the critical cone is $\{0\}$, so the second-order sufficient condition is satisfied vacuously. Feasible motion still exists: $\mathbf p=(1,2)^T$ is tangent to the equality and enters the inequality interior, but raises the objective to first order.

This particular example also admits a global certificate. On its positive log domain, $4-x_1x_2\le0$ is equivalent to $4/x_1-x_2\le0$, a convex inequality; its second derivative in $x_1$ is $8/x_1^3>0$. The equality is affine, and the objective Hessian is positive definite throughout the domain: its diagonal entries exceed 2 and 3, with off-diagonal entry 1, so its determinant exceeds 5. The problem therefore has a convex feasible set and a strictly convex objective even though the original product constraint function is not convex. The local minimum is the unique global minimum. This illustrates the representation distinction in [Convexity Revisited](Convexity.ipynb), rather than applying a failed structural test as a converse.

Sources: Biegler (2010), Eqs. (4.16)–(4.24), p.72; Theorem 4.3, p.65, and Theorem 4.18, p.81. The equivalent-constraint/global certificate and exact multipliers are independently derived from the displayed model.
Objective-value source correction: the displayed model has constant +9, yielding approximately -0.874224 without constraints and 0.157862 with both constraints. The book’s -2.8742 and -1.8421 are two units lower and correspond to replacing that constant by +7. The additive constant changes values but not gradients, minimizers, multipliers or curvature. We keep the displayed +9 model and its independently evaluated values.


### Active-branch and Hessian details for this same model

If the equality is retained but the inequality is assumed slack, set $u=0$ and $x_2=2x_1$. The restricted objective is

$$f(x_1,2x_1)=9x_1^2-18x_1+9-\ln2-2\ln x_1.$$

Stationarity gives $18x_1-18-2/x_1=0$, or $9x_1^2-9x_1-1=0$. Its positive root is $(9+\sqrt{117})/18\approx1.100925$, with $x_2\approx2.201850$. But $g=4-2x_1^2\approx1.575927>0$, so this branch is infeasible. The active branch instead enforces $4-2x_1^2=0$, and the positive domain selects the exact point derived above. A root is accepted only after domain, primal feasibility, multiplier sign and complementarity checks.

The Lagrangian Hessian is a sum of scalar-component Hessians,

$$\nabla^2_{xx}L=\nabla^2f+u\nabla^2g+v\nabla^2h
=\begin{bmatrix}2+1/x_1^2&1-u\\1-u&3+1/x_2^2\end{bmatrix}.$$

At the exact constrained point, the diagonal entries are $2.5$ and $3.125$, and the off-diagonal is $1-u^*=(9\sqrt2-13)/4\approx-0.068019$. Biegler Eq.(4.24) p.72 prints a positive off-diagonal. The errata fixes the diagonal signs but leaves that further sign error. The critical-direction matrix

$$\begin{bmatrix}-2\sqrt2&-\sqrt2\\2&-1\end{bmatrix}$$

has determinant $4\sqrt2$, so its null space is zero. It describes directions tangent to the equality and the positively weighted active inequality. It does not describe every feasible direction: $(1,2)^T$ is an inward feasible rail direction and increases the objective to first order.

![The rail, inequality tangent and feasible inward step](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-directions.png)

**Printed orientation discrepancy.** Biegler Eq.(4.9) p.70 uses transposed vector-constraint gradients, while Eqs.(4.17)–(4.18) p.72 use columns for the same stationarity components. The course consistently defines $\mathbf J_g$ with constraint rows ($r\times n$), so $\mathbf J_g\mathbf p$ is a constraint-change vector and $\mathbf J_g^T\mathbf u$ is an $n$-component gradient contribution. A single scalar-constraint gradient is a column. This resolves the dimensions without adopting two definitions of the same symbol.
