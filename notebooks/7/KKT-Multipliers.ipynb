## Active Sets

For $\min f(\mathbf{x})$ subject to $\mathbf{g}(\mathbf{x})\le\mathbf{0}$ and $\mathbf{h}(\mathbf{x})=\mathbf{0}$, define equality indices $\mathcal{E}=\{1,\ldots,m\}$ and inequality indices $\mathcal{I}=\{1,\ldots,r\}$. The active **inequality** index set at a feasible point is

$$\mathcal{A}_I(\mathbf{x})=\{i\in\mathcal{I}:g_i(\mathbf{x})=0\}.$$

If both types are collected into one set, use distinct constraint labels:

$$\mathcal{A}(\mathbf{x})=\{(h,j):j\in\mathcal{E}\}\cup\{(g,i):i\in\mathcal{A}_I(\mathbf{x})\}.$$

Equality $h_1$ and inequality $g_1$ are different constraints, even though their component indices are both 1. The inactive inequality indices are $\mathcal{I}\setminus\mathcal{A}_I$, rather than a difference with untagged equality indices. All equalities belong to the combined active set at every feasible point.

**Why an inactive constraint drops out locally.** Suppose $g_i(\mathbf{x}^*)<0$ and $g_i$ is continuous. There is a neighborhood of $\mathbf{x}^*$ on which $g_i<0$ throughout. For finitely many inactive inequalities, take the intersection of these neighborhoods. Removing them leaves exactly the same feasible comparisons in that neighborhood. Their KKT multipliers are zero by complementarity, so they contribute nothing to gradient balance either. This is a local argument, not permission to remove them from the optimization model.

![An inactive constraint leaves the local neighborhood unchanged but excludes a lower distant value](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-local-inactive.png)

The prepared example uses $f(x)=(x^2-1)^2+0.1(3x-x^3)$ and $g=-x\le0$. Its derivative is $(x^2-1)(4x-0.3)$. At $x=1$, $f''=7.4>0$, $g=-1$ and $f=0.2$. Every point in $(0.7,1.3)$ satisfies the inequality, so it is irrelevant to that local test. Removing it globally admits the other strict local minimum $x=-1$, with value $-0.2$. A constraint that is inactive at the current solution can still exclude a better distant solution.

Source: Biegler p.72 for the active/local characterization; explicit index spaces, continuity proof and polynomial example are independently specified here.

## Sensitivity Analysis

### A smooth local branch and a moving constraint

Choose $\hat\imath\in\mathcal{A}_I(\mathbf{x}^*)$ and replace just one inequality by

$$g_{\hat\imath}(\mathbf{x})+\epsilon\le0.$$

Positive $\epsilon$ tightens the right-hand side to $-\epsilon$; negative $\epsilon$ relaxes it. Let $\mathbf{x}^{\epsilon}$ be a nearby **local minimizer branch**, with $\mathbf{x}^0=\mathbf{x}^*$, and put $V(\epsilon)=f(\mathbf{x}^{\epsilon})$. This is a local branch value. It need not select the global minimum for every perturbed problem.

![A fixed equality rail and a moving active inequality fence](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-perturbation.png)

The drawing uses $f=x_1^2+x_2^2$, $g=b-x_1\le0$, $h=x_2-x_1=0$, and $b=1$. The exact solution branch is $\mathbf{x}^{\epsilon}=(b+\epsilon,b+\epsilon)^T$ for $b+\epsilon>0$. It has $V(\epsilon)=2(b+\epsilon)^2$, $V'(0)=4b$, inequality multiplier $u^*=4b$ and equality multiplier $v^*=-2b$.

**Hypotheses for the following Taylor proof.** The data are continuously differentiable; $\mathbf{x}^{\epsilon}$ is differentiable at zero; the same inequalities remain active for small positive and negative $\epsilon$. Thus

$$\Delta\mathbf{x}:=\mathbf{x}^{\epsilon}-\mathbf{x}^*=\epsilon\mathbf{x}'(0)+o(\epsilon)=O(\epsilon).$$

Taylor remainders are therefore $o(\epsilon)$. The active-set statement alone does not justify dividing an uncontrolled approximation by $\epsilon$.

1. Equalities are satisfied at both points:

$$\mathbf{0}=\mathbf{h}(\mathbf{x}^{\epsilon})-\mathbf{h}(\mathbf{x}^*)=\mathbf{J}_h(\mathbf{x}^*)\Delta\mathbf{x}+o(\epsilon).$$

2. Every unperturbed active inequality is still zero:

$$0=g_i(\mathbf{x}^{\epsilon})-g_i(\mathbf{x}^*)=\nabla g_i(\mathbf{x}^*)^T\Delta\mathbf{x}+o(\epsilon),\qquad i\in\mathcal{A}_I\setminus\{\hat\imath\}.$$

3. The selected one stays active at its changed bound:

$$-\epsilon=g_{\hat\imath}(\mathbf{x}^{\epsilon})-g_{\hat\imath}(\mathbf{x}^*)=\nabla g_{\hat\imath}(\mathbf{x}^*)^T\Delta\mathbf{x}+o(\epsilon).$$

### Full objective substitution

![Constraint changes multiplied by minus their multipliers leave one first-order term](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-proof-contributions.png)

At the unperturbed KKT point,

$$\nabla f=-\mathbf{J}_h^T\mathbf{v}^*-\mathbf{J}_g^T\mathbf{u}^*.$$

Use the objective expansion and substitute this identity:

$$\begin{aligned}
V(\epsilon)-V(0)&=\nabla f(\mathbf{x}^*)^T\Delta\mathbf{x}+o(\epsilon)\\
&=-\sum_{j\in\mathcal E}v_j^*\nabla h_j^T\Delta\mathbf{x}-\sum_{i\in\mathcal I}u_i^*\nabla g_i^T\Delta\mathbf{x}+o(\epsilon)\\
&=-\sum_{j\in\mathcal E}v_j^*\bigl(h_j(\mathbf{x}^{\epsilon})-h_j(\mathbf{x}^*)\bigr)
\\&\quad-\sum_{i\in\mathcal I}u_i^*\bigl(g_i(\mathbf{x}^{\epsilon})-g_i(\mathbf{x}^*)\bigr)+o(\epsilon)\\
&=u_{\hat\imath}^*\epsilon+o(\epsilon).
\end{aligned}$$

All equality differences vanish. Inactive inequalities have $u_i^*=0$. Untouched active inequalities have zero constraint difference. The selected constraint has difference $-\epsilon$, and its contribution is $-u_{\hat\imath}^*(-\epsilon)$. Hence

$$V'(0)=\lim_{\epsilon\to0}\frac{V(\epsilon)-V(0)}{\epsilon}=u_{\hat\imath}^*.$$

This is the sign convention for **$g+\epsilon\le0$**. If instead the constraint is written $g\le b$, increasing $b$ relaxes it and the derivative with respect to $b$ is $-u^*$. For an equality written $h+\eta=0$, the same substitution gives $\partial V/\partial\eta=v^*$. With a differentiable multi-parameter branch,

$$dV=\mathbf{u}^{*T}d\boldsymbol\epsilon+\mathbf{v}^{*T}d\boldsymbol\eta.$$

Rescaling $g$ to $c g$, $c>0$, changes the multiplier to $u/c$. Its units and numerical magnitude depend on the constraint's scaling; the product of multiplier and the corresponding constraint perturbation is invariant. A large multiplier is not comparable across arbitrary scalings.

Source: Biegler p.73 for the Taylor argument. The branch/remainder scope and sign/scaling extensions are stated explicitly; Rockafellar Thm.9.1 p.224 distinguishes an ordinary derivative from a general subgradient interpretation.

### Strong activity, weak activity, and strict complementarity

For a **chosen KKT multiplier vector**, the active inequality indices split into

$$\mathcal{A}_s=\{i\in\mathcal{A}_I:u_i^*>0\},\qquad\mathcal{A}_w=\{i\in\mathcal{A}_I:u_i^*=0\}.$$

| Inequality state | Constraint value | Multiplier |
| --- | --- | --- |
| inactive | $g_i(\mathbf{x}^*)<0$ | $u_i^*=0$ |
| strongly active | $g_i(\mathbf{x}^*)=0$ | $u_i^*>0$ |
| weakly active | $g_i(\mathbf{x}^*)=0$ | $u_i^*=0$ |

Strict complementarity means every active inequality is strongly active, equivalently $\mathcal A_w$ is empty. If there are no active inequalities it holds vacuously. Equality multipliers are unrestricted and are not classified by this inequality definition. Nonunique multiplier vectors can yield different strong/weak classifications for the same primal point; uniqueness is a separate regularity question.

![Three lower-bound examples for classifying active states](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-active-states.png)

For $\min(x-a)^2$ subject to $-x\le0$, $x^*=\max(a,0)$ and stationarity is $2(x^*-a)-u^*=0$. At $a=-1$, the bound is strongly active with $u^*=2$. At $a=0$, it is weakly active with $u^*=0$. At $a=1$, it is inactive with $u^*=0$. Zero multiplier does not imply slack.

Source: Biegler Def.4.6 and the preceding sets, p.73; the explicit prepared examples supply the distinctions.

### Example 4.7: enumerate and test the bound states

$$\min_x x^2\quad\text{s.t.}\quad \tfrac12-x\le0,\quad x-1\le0.$$

The Lagrangian and KKT conditions are

$$L=x^2+u_L(\tfrac12-x)+u_U(x-1),$$

$$2x-u_L+u_U=0,\quad u_L(\tfrac12-x)=0,\quad u_U(x-1)=0,$$

$$\tfrac12\le x\le1,\qquad u_L,u_U\ge0.$$

| Active subset | Stationarity/active equations | Result |
| --- | --- | --- |
| neither | $u_L=u_U=0$, $2x=0$ | $x=0$ violates the lower bound |
| lower only | $x=1/2$, $u_U=0$ | $u_L=1$, accepted |
| upper only | $x=1$, $u_L=0$ | $u_U=-2$, violates the multiplier sign |
| both | $x=1/2$ and $x=1$ | infeasible |

Thus $x^*=1/2$, $u_L^*=1$, $u_U^*=0$. The lower bound is strongly active and strict complementarity holds. The objective and bounds are convex, so this candidate is the global minimum. Enumeration considers up to $2^r$ subsets; infeasible or sign-inconsistent branches are rejected, rather than accepted just because some stationarity equations have a root. Active-set and interior-point methods avoid exhaustive enumeration.

![Decision-space geometry of the strong and weak bound examples](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-bound-examples.png)

Perturb the lower bound to $1/2+\epsilon$. For $-1/2<\epsilon<1/2$, it remains active:

$$x^{\epsilon}=\tfrac12+\epsilon,\qquad V_s(\epsilon)=\tfrac14+\epsilon+\epsilon^2,\qquad V_s'(0)=1=u_L^*.$$

The exact full value function is

$$V_s(\epsilon)=\begin{cases}0,&\epsilon\le-1/2,\\(1/2+\epsilon)^2,&-1/2<\epsilon\le1/2,\\+\infty,&\epsilon>1/2.\end{cases}$$

For $\epsilon<-1/2$, the lower bound is inactive; at $\epsilon=1/2$ both bounds coincide and the feasible set has one point. Beyond that the problem is infeasible. The original slope 1 predicts only the local linear term; for $\epsilon=0.1$ the true increase is $0.11$, not $0.10$. Continuing that old line far enough can even predict a negative value for a squared objective.

### Weak-activity variation: derive the zero multiplier

Now use $\min x^2$ subject to $0\le x\le1$. Write

$$L=x^2-u_Lx+u_U(x-1),\qquad 2x-u_L+u_U=0.$$

At $x^*=0$, the upper bound is slack, so complementarity gives $u_U^*=0$. The **full** stationarity equation then gives $u_L^*=0$. The lower bound is active but weak, and strict complementarity fails. This derives the result without assuming $u_L=0$ in advance. The 2018 scan's case-ii premise states that value; keeping it in the reference as a source reading does not make it a necessary premise for the derivation.

Perturb the lower bound to $x\ge\epsilon$:

$$x^{\epsilon}=\max(\epsilon,0),\qquad
V_w(\epsilon)=\begin{cases}0,&\epsilon\le0,\\\epsilon^2,&0<\epsilon\le1,\\+\infty,&\epsilon>1.\end{cases}$$

Tightening to $\epsilon>0$ changes the value by $\epsilon^2$; loosening keeps the unconstrained minimizer and changes the value by zero. Both one-sided first derivatives at zero are zero, matching the multiplier. Their second-order behavior differs.

![Exact optimum-value functions compared with the original multiplier predictions](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/multiplier-sensitivity.png)

An active-set change requires rechecking the smooth-branch interpretation; it does **not** automatically destroy the value derivative. Both value functions are continuously differentiable at the slack-to-active transitions shown in the plot ($\epsilon=-1/2$ and $\epsilon=0$), with a change in the second derivative. Their feasibility-domain endpoints are separate: perturbing beyond $1/2$ or $1$, respectively, makes the model infeasible. In the weak example the solution branch has different one-sided derivatives, but the value still has derivative zero. For a finite perturbation, recompute the relevant active state and multiplier rather than extrapolating the old slope indefinitely.

Source: Biegler Example 4.7 and variation, pp.73–74. Case enumeration, complete parameter domains and the two-sided qualifications are explicit here.

### Computing a branch derivative with the active KKT matrix

For twice continuously differentiable data, collect equalities and selected active inequalities into constraint rows $\mathbf J_A$, with multipliers $\boldsymbol\lambda$. A constant right-hand-side perturbation changes one active equation but not its gradient. Differentiating the active stationarity/equality equations of a smooth branch gives

$$\begin{bmatrix}\mathbf W&\mathbf J_A^T\\\mathbf J_A&\mathbf0\end{bmatrix}
\begin{bmatrix}\mathbf x'(0)\\\boldsymbol\lambda'(0)\end{bmatrix}
=\begin{bmatrix}\mathbf0\\-\mathbf e_{\hat\imath}\end{bmatrix},\qquad
\mathbf W=\nabla^2_{xx}L(\mathbf x^*,\boldsymbol\lambda^*).$$

Here $\mathbf e_{\hat\imath}$ is the unit vector in the stacked row corresponding to the perturbed inequality. Its row position includes the equality rows; it is not an unadjusted original inequality index.

This is a derivative of the active KKT equations, not an optimization algorithm. A nonsingular matrix gives one branch derivative. Nocedal & Wright (2006), Appendix A.2, Theorem A.2, p.631, gives the implicit function theorem used here: it supplies a nearby smooth **root** branch when this active KKT Jacobian is nonsingular; feasibility, multiplier signs and a local sufficiency test are still needed to identify it as the intended minimizer branch. Strictly positive active multipliers and strictly slack inactive constraints persist under small continuous changes; weak activity needs special care.

One sufficient nonsingularity test is full row rank of $\mathbf J_A$ and positive definiteness of $\mathbf W$ on its null space. To verify it, consider a homogeneous solution $(\mathbf p,\mathbf q)$:

$$\mathbf W\mathbf p+\mathbf J_A^T\mathbf q=\mathbf0,\qquad\mathbf J_A\mathbf p=\mathbf0.$$

Multiplying the first equation by $\mathbf p^T$ gives $\mathbf p^T\mathbf W\mathbf p=0$. Positive curvature on the null space forces $\mathbf p=\mathbf0$. Then $\mathbf J_A^T\mathbf q=\mathbf0$, and full row rank forces $\mathbf q=\mathbf0$. If the active Jacobian is square and nonsingular, its null space is already zero. This explains why the moving-fence/rail toy problem has a well-defined branch even before a nonzero tangent-curvature direction is available.

![Constraint-row movement and transposed-multiplier balance are the two actions in the bordered system](https://raw.githubusercontent.com/ndcbe/optimization/main/media/figures/kkt-jacobian-roles.png)

The matrix construction and nonsingularity argument are independently derived from the displayed KKT equations; the extended reduced-Hessian treatment is in [Second Order Optimality Conditions](Second-Order.ipynb).

### When the value has two one-sided slopes

Consider the convex problem

$$\min_x x\quad\text{s.t.}\quad\epsilon-x\le0,\quad -x\le0.$$

Its exact value is $V(\epsilon)=\max(\epsilon,0)$. At zero both constraints are active and stationarity is $1-u_1-u_2=0$ with nonnegative multipliers. Every pair $(u_1,u_2)=(t,1-t)$, $0\le t\le1$, is valid. The same primal point therefore has nonunique multipliers. The right value derivative is 1, the left derivative is 0, and there is no ordinary two-sided derivative. The selected constraint multiplier $u_1\in[0,1]$ is a subgradient, not an arbitrary choice of a common slope.

This example also shows why strong/weak labels can depend on the multiplier vector: $(1,0)$ and $(0,1)$ exchange which active constraint is weak. The two gradients are both $-1$, so LICQ fails, even though a direction $d>0$ strictly decreases both active inequality linearizations.

Rockafellar Thm.9.1 p.224 states a regular-subgradient interpretation under its hypotheses; ordinary partial derivatives follow when the relevant value function is differentiable. Thm.9.3 p.225 gives the convex directional/subgradient relation under its stated existence and lower-semicontinuity hypotheses. In that convex setting, the right and left coordinate slopes are the largest and smallest compatible coordinate multipliers. Those statements should not be read as a global smooth sensitivity theorem for arbitrary nonconvex problems.

The local Taylor proof, the weak squared-value example and this kink example serve different purposes: a smooth branch gives a derivative; a branch switch can leave the value differentiable; nonunique multipliers can describe a range of one-sided sensitivities.

## Multipliers in Pyomo

Reference: [Pyomo Suffix documentation](https://pyomo.readthedocs.io/en/stable/pyomo_modeling_components/Suffixes.html), especially importing solver data and exporting warm-start information. `IMPORT` retrieves solver-supported suffixes, `EXPORT` sends them, and `IMPORT_EXPORT` does both. Names and support depend on the solver/interface. The existing example below uses the Ipopt NL interface: `dual` for constraint multipliers and `ipopt_zL_out`/`ipopt_zU_out` for variable-bound multipliers. The `_in` suffixes provide the bound-multiplier warm start. Check termination before interpreting any stored suffix, and check the model's stationarity/sign convention when comparing it with $g\le0$ mathematics.

The warm start below solves the same model again, using the first primal/dual point. It is a demonstration of passing initialization, not a guarantee that warm starting always reduces iterations or selects a global minimum.

$$
\begin{align} \min_{x_1,...,x_4} \quad & x_1 \cdot x_4 \cdot (x_1 + x_2 + x_3) + x_3 \\
\mathrm{s.t.} \quad & x_1 \cdot x_2 \cdot x_3 \cdot x_4 \geq 25 \\
 & x_1^2 + x_2^2 + x_3^2 + x_4^2 = 40
\end{align}
$$

````python
import pyomo.environ as pyo

# Example from
# https://pyomo.readthedocs.io/en/stable/pyomo_modeling_components/Suffixes.html#exporting-suffix-data

model = pyo.ConcreteModel()
model.x1 = pyo.Var(bounds=(1, 5), initialize=1.0)
model.x2 = pyo.Var(bounds=(1, 5), initialize=5.0)
model.x3 = pyo.Var(bounds=(1, 5), initialize=5.0)
model.x4 = pyo.Var(bounds=(1, 5), initialize=1.0)
model.obj = pyo.Objective(
    expr=model.x1 * model.x4 * (model.x1 + model.x2 + model.x3) + model.x3
)
model.inequality = pyo.Constraint(
    expr=model.x1 * model.x2 * model.x3 * model.x4 >= 25.0
)
model.equality = pyo.Constraint(
    expr=model.x1**2 + model.x2**2 + model.x3**2 + model.x4**2 == 40.0
)

### Declare all suffixes
# Ipopt bound multipliers (obtained from solution)
model.ipopt_zL_out = pyo.Suffix(direction=pyo.Suffix.IMPORT)
model.ipopt_zU_out = pyo.Suffix(direction=pyo.Suffix.IMPORT)
# Ipopt bound multipliers (sent to solver)
model.ipopt_zL_in = pyo.Suffix(direction=pyo.Suffix.EXPORT)
model.ipopt_zU_in = pyo.Suffix(direction=pyo.Suffix.EXPORT)
# Obtain dual solutions from first solve and send to warm start
model.dual = pyo.Suffix(direction=pyo.Suffix.IMPORT_EXPORT)

ipopt = pyo.SolverFactory("ipopt")
````

### Solve without warm starting

````python
results = ipopt.solve(model, tee=True)
assert pyo.check_optimal_termination(results), (
    f"Solve failed: status={results.solver.status}, "
    f"termination={results.solver.termination_condition}"
)
````

### Solve with warm starting

````python
### Set Ipopt options for warm-start
# The current values on the ipopt_zU_out and ipopt_zL_out suffixes will
# be used as initial conditions for the bound multipliers to solve the
# new problem
model.ipopt_zL_in.update(model.ipopt_zL_out)
model.ipopt_zU_in.update(model.ipopt_zU_out)
ipopt.options["warm_start_init_point"] = "yes"
ipopt.options["warm_start_bound_push"] = 1e-6
ipopt.options["warm_start_mult_bound_push"] = 1e-6
ipopt.options["mu_init"] = 1e-6

results = ipopt.solve(model, tee=True)
assert pyo.check_optimal_termination(results), (
    f"Solve failed: status={results.solver.status}, "
    f"termination={results.solver.termination_condition}"
)
````

### Translate solver suffixes and check gradient balance

For this minimization model and Ipopt NL interface, use the canonical inequality $g=25-x_1x_2x_3x_4\le0$ and equality $h=\sum_i x_i^2-40=0$. The imported inequality dual is $u$, and the equality multiplier in $L=f+u g+v h$ is the negative of the imported equality dual. The canonical lower-bound multiplier is `ipopt_zL_out`; the canonical upper-bound multiplier is the negative of the signed `ipopt_zU_out` suffix. Lower-bound multipliers enter stationarity with a minus sign, upper-bound multipliers with a plus sign. Keep the raw signed suffix values when passing a warm start back to the solver. The calculation below checks these signs against independently written analytic gradients, rather than assuming that every solver/interface uses the same suffix convention.

A small residual establishes approximate KKT consistency for this returned point, not a global-minimum certificate for the nonconvex model. The warm-start demonstration repeats the same problem; timing and iteration counts can change with the solver version.

The existing four-variable model has inactive upper bounds. A separate one-variable check, $\min -x$ with $0\le x\le1$, makes the upper bound strongly active and verifies its sign with a nonzero multiplier.


````python
import numpy as np

x = np.array([pyo.value(model.x1), pyo.value(model.x2),
              pyo.value(model.x3), pyo.value(model.x4)])
variables = [model.x1, model.x2, model.x3, model.x4]
u = float(model.dual[model.inequality])
v = -float(model.dual[model.equality])
z_lower = np.array([float(model.ipopt_zL_out.get(var, 0.0)) for var in variables])
z_upper = -np.array([float(model.ipopt_zU_out.get(var, 0.0)) for var in variables])

# Analytic derivatives of the displayed objective and canonical constraints.
gradient_f = np.array([x[3] * (2*x[0] + x[1] + x[2]),
                       x[0]*x[3], x[0]*x[3] + 1,
                       x[0]*(x[0] + x[1] + x[2])])
gradient_g = -np.prod(x) / x
gradient_h = 2*x
stationarity = gradient_f + u*gradient_g + v*gradient_h - z_lower + z_upper
primal_residual = max(0.0, 25-np.prod(x), abs(np.dot(x, x)-40),
                      float(np.max(1-x)), float(np.max(x-5)))
complementarity = max(abs(u*(25-np.prod(x))),
                      float(np.max(np.abs(z_lower*(1-x)))),
                      float(np.max(np.abs(z_upper*(x-5)))))
assert np.max(np.abs(stationarity)) < 1e-5
assert primal_residual < 1e-5 and complementarity < 1e-5
assert u >= -1e-7 and np.min(z_lower) >= -1e-7 and np.min(z_upper) >= -1e-7

print("Primal point:", np.round(x, 8))
print(f"Objective: {pyo.value(model.obj):.8f}")
print(f"Canonical inequality multiplier u: {u:.8f}")
print(f"Canonical equality multiplier v: {v:.8f}")
print("Lower-bound multipliers:", np.round(z_lower, 8))
print("Upper-bound multipliers:", np.round(z_upper, 8))
print(f"Max stationarity residual: {np.max(np.abs(stationarity)):.3e}")
print(f"Max primal residual: {primal_residual:.3e}")
print(f"Max complementarity residual: {complementarity:.3e}")

# A nonzero upper-bound multiplier checks the signed suffix conversion.
probe = pyo.ConcreteModel()
probe.x = pyo.Var(bounds=(0, 1), initialize=0.5)
probe.obj = pyo.Objective(expr=-probe.x)
probe.ipopt_zL_out = pyo.Suffix(direction=pyo.Suffix.IMPORT)
probe.ipopt_zU_out = pyo.Suffix(direction=pyo.Suffix.IMPORT)
probe_result = pyo.SolverFactory("ipopt").solve(probe)
assert pyo.check_optimal_termination(probe_result)
raw_upper = float(probe.ipopt_zU_out[probe.x])
canonical_upper = -raw_upper
canonical_lower = float(probe.ipopt_zL_out[probe.x])
probe_stationarity = -1 - canonical_lower + canonical_upper
assert raw_upper < -0.5 and canonical_upper > 0.5
assert abs(probe_stationarity) < 1e-6
assert abs(pyo.value(probe.x)-1) < 1e-6
print(f"Active-upper-bound raw suffix: {raw_upper:.8f}")
print(f"Active-upper-bound canonical multiplier: {canonical_upper:.8f}")
print(f"Active-upper-bound stationarity residual: {abs(probe_stationarity):.3e}")
````
